```python
# ---------------------------------------------
# Name      : SWATHI
# Roll No   : 24BAD122
# Experiment: Banking Customer Data Analysis
# ---------------------------------------------

# Import Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load Dataset (Download from Kaggle and place in folder)
data = pd.read_csv("marketing_campaign.csv", sep='\t')

# -----------------------------
# Examine Dataset Structure
# -----------------------------
print("First 5 Rows")
print(data.head())

print("\nDataset Info")
print(data.info())

print("\nDataset Shape:", data.shape)

# -----------------------------
# Check Missing Values
# -----------------------------
print("\nMissing Values")
print(data.isnull().sum())

# -----------------------------
# Column Types
# -----------------------------
print("\nColumn Data Types")
print(data.dtypes)

# -----------------------------
# Create Age Column
# -----------------------------
current_year = 2024
data['Age'] = current_year - data['Year_Birth']

# -----------------------------
# Create Total Spending Column
# -----------------------------
data['Total_Spending'] = (
    data['MntWines'] +
    data['MntFruits'] +
    data['MntMeatProducts'] +
    data['MntFishProducts'] +
    data['MntSweetProducts'] +
    data['MntGoldProds']
)

# -----------------------------
# Handle Missing Income Values
# -----------------------------
data['Income'].fillna(data['Income'].median(), inplace=True)

# -----------------------------
# Visualization – Bar Plot (Average Spending by Education)
# -----------------------------
plt.figure()
sns.barplot(x='Education', y='Total_Spending', data=data)
plt.xticks(rotation=45)
plt.title("Average Spending by Education")
plt.show()

# -----------------------------
# Box Plot – Age Distribution
# -----------------------------
plt.figure()
sns.boxplot(y=data['Age'])
plt.title("Age Distribution")
plt.show()

# -----------------------------
# Box Plot – Income Distribution
# -----------------------------
plt.figure()
sns.boxplot(y=data['Income'])
plt.title("Income Distribution")
plt.show()

# -----------------------------
# Box Plot – Spending Distribution
# -----------------------------
plt.figure()
sns.boxplot(y=data['Total_Spending'])
plt.title("Total Spending Distribution")
plt.show()
```
